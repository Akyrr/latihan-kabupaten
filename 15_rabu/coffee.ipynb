```python
def split_data(X, y, size=0.8):
    np.random.seed(42)

    x_train, x_test = [], []
    y_train, y_test = [], []
    for class_value in np.unique(y):
        class_idx = np.where(y == class_value)[0]

        idx = np.random.permutation(class_idx)
        split = int(len(idx) * size)

        x_train.append(X.iloc[idx[:split]])
        x_test.append(X.iloc[idx[split:]])
        y_train.append(y.iloc[idx[:split]])
        y_test.append(y.iloc[idx[split:]])

    X_train = pd.concat(x_train).reset_index(drop = True)
    X_test = pd.concat(x_test).reset_index(drop = True)
    y_train = pd.concat(y_train).reset_index(drop = True)
    y_test = pd.concat(y_test).reset_index(drop = True)

    return X_train, X_test, y_train, y_test
```

```python
X_train, X_test, y_train, y_test = split_data(X, y)
```

```python
print(X_train.shape)
print(y_train.shape)
print(X_test.shape)
print(y_test.shape)
```

```python
class KNN:
    def __init__(self, k, p):
        self.k = k
        self.p = p
    
    def train(self, X, y):
        self.X_train = X
        self.y_train = y

    def distance(self, x1, x2):
        if self.p == 1:
            return np.sum(np.abs(x1 - x2))
        else:
            return np.sqrt(np.sum((x1 - x2)**2))
    
    def predict(self, X):
        y_pred = [self._predict(x) for x in X]
        return np.array(y_pred)
    
    def _predict(self, x):
        distance_i = [self.distance(x, X_train) for X_train in self.X_train]
        best_k  = np.argsort(distance_i)[:self.k]
        label_k = [self.y_train[i] for i in best_k]
        result = max(set(label_k), key=label_k.count)
        return result
```

```python
best_acc = 0
best_k = {}

for k in [3,5,7,9,11,13]:
    for p in [1,2]:
        model = KNN(k, p)
        model.train(X_train.values, y_train.values)
        pred = model.predict(X_test.values)
        acc = np.mean(pred == y_test.values)

        distance = 'manhattan' if p == 1 else 'euclidean'
        print(f'K: {k} - distance: {distance} - acc: {acc}')

        if acc > best_acc:
            best_acc = acc
            best_k = {'k': k, 'p':p}
    distance_label = 'manhattan' if p == 1 else 'euclidean'
print(f'best k {best_k.get('k', 'N/A')} - distance: {distance_label} - akurasi: {best_acc}')
```

```python
y_pred_train = model.predict(X_train.values)
y_pred_test = model.predict(X_test.values)
```

```python
def evaluate(y_true, y_pred, title='Train'):
    y_true = np.array(y_true)
    y_pred = np.array(y_pred)

    classes = np.unique(y_true)
    n_classes = len(classes)

    accuracy = np.mean(y_true == y_pred)
    precision = []
    recall = []
    f1 = []

    for i in classes:
        tp = np.sum((y_true == i) & (y_pred == i))
        fp = np.sum((y_true != i) & (y_pred == i))
        fn = np.sum((y_true == i) & (y_pred != i))

        p = tp / (tp + fp) if (tp + fp) > 0 else 0
        r = tp / (tp + fn) if (tp + fn) > 0 else 0
        f = 2 * (p * r) / (p + r) if (p + r) > 0 else 0

        precision.append(p)
        recall.append(r)
        f1.append(f)

    precision = np.mean(precision)
    recall = np.mean(recall)
    f1 = np.mean(f1)

    print(f'Evaluasi {title}')
    print(f'Akurasi     : {accuracy:.4f}')
    print(f'Precision   : {precision:.4f}')
    print(f'Recall      : {recall:.4f}')
    print(f'F1          : {f1:.4f}')
```

```python
evaluate(y_train, y_pred_train)
```

```python
evaluate(y_test, y_pred_test, title='Test')
```

```python
def confusion_matrix(y_true, y_pred, title):
    clases = np.unique(y_true)
    n_classes = len(clases)
    label = coffee['Target'].map({0:'Low', 1:'Medium', 2:'High'})

    cm = np.zeros((n_classes, n_classes))

    for i, c_true in enumerate(clases):
        for j, c_pred in enumerate(clases):
            cm[i, j] = np.sum((y_true == c_true) & (y_pred == c_pred))

    sns.heatmap(cm, annot=True, fmt='.1f' ,cmap='Blues', xticklabels=label.unique(), yticklabels=label.unique())
    plt.title(f'Evaluasi {title}')
    plt.xlabel('Prediksi')
    plt.ylabel('Aktual')
    plt.show()
```

```python
confusion_matrix(y_train, y_pred_train, 'Train')
```

```python
confusion_matrix(y_test, y_pred_test, 'Test')
```

```python
def incorrect_point(y_true, y_pred, X, title):
    incorrect = y_true != y_pred

    plt.figure(figsize=(10,8))
    plt.scatter(X[incorrect, 0], X[incorrect, 1], color='red', label='Salah')
    plt.scatter(X[~incorrect, 0], X[~incorrect, 1], color='blue', label='Benar')
    plt.title(f'Poin Kesalahan {title}')
    plt.legend()
    plt.grid()
    plt.show()
```
