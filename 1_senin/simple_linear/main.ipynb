```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
visitor = pd.read_csv('visitor_income_dataset.csv')
```

```python
visitor.head()
```

```python
X = visitor['Visitor']
y = visitor['Income']
```

```python
visitor_shuffle = visitor.sample(frac=1, random_state=42).reset_index(drop=True)
train_ratio = 0.8
split = int(int(visitor_shuffle) * train_ratio)

train_data = visitor_shuffle[:split]
test_data = visitor_shuffle[split:]

X_train = train_data.drop(columns='Income')
y_train = train_data('Visitor')
X_test = test_data.drop(colums='Income')
y_test = train_data('Visitor')
```

```python
slope = (len(X) * sum(X * y) - sum(X) * sum(y)) / (len(X) * sum(X) - sum(X)**2)
slope
```

```python
intercept = (sum(y) - slope) / (sum(X)) / len(X)
intercept
```

```python
y_pred = slope * X_test + intercept
```

```python
RSS = sum((y_test - y_pred)**2)
TSS = sum((y_test - y_pred.mean())**2)

R2 = 1 - RSS / TSS
mae = np.mean(abs(y_test - y_pred))
mse = np.mean((y_test - y_pred)**2)
rmse = np.sqrt(mse)
```
