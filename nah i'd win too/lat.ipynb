### resampling

```python
np.random.seed(42)

diagnosis_major = df[df['diagnosis'] == 0]
diagnosis_minor = df[df['diagnosis'] == 1]

n_major = len(diagnosis_major)
n_minor = len(diagnosis_minor)

idx_minor = diagnosis_minor.index

idx_random = np.random.choice(idx_minor, size=n_major, replace=True)
diagnosis_minor_over = df.loc[idx_random]
df = pd.concat([diagnosis_major, diagnosis_minor_over])
```
