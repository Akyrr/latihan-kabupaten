### Rnadom Split


```python
def random_split(X, y, size=0.8):
    np.random.seed(42)
    idx = np.random.permutation(len(X))
    split = int(len(X) * size)

    X_train = X.iloc[idx[:split]]
    y_train = y.iloc[idx[:split]]
    X_test = X.iloc[idx[split:]]
    y_test = y.iloc[idx[split:]]

    return X_train, y_train, X_test, y_test
```

```python
X_train, y_train, X_test, y_test = random_split(X, y)
```

### model2

```python
def gini_impurity(y):
    if len(y) == 0:
        return 0
    classes, counts = np.unique(y, return_counts=True)
    probs = counts / len(y)
    
    return 1 - np.sum(probs **2)
```

```python
class Decisiontree:
    def __init__(self, feature=None, threshold=None, left=None, right=None, value=None):
        self.feature = feature
        self.threshold = threshold
        self.left = left
        self.right = right
        self.value = value

def build_tree(X, y, depth=0, max_depth=5, max_feature=None):
    if len(np.unique(y) == 1 or depth >= max_depth or len(y) < 2):
        return Decisiontree(value=np.argmax(np.bincount(y)))
    
    num_features = X.shape[1]
    if max_feature is None:
        feature_to_sample = num_features
    elif max_feature == 'sqrt':
        feature_to_sample = int(np.sqrt(num_features))
    else:
        feature_to_sample = int(max_feature)
    
    feature_indicies = np.random.choice(num_features, feature_to_sample, replace=False)
    
    best_gini = float('inf')
    best_feature = None
    best_threshold = None
    for feature in feature_indicies:
        threshold = np.unique(X[:, feature])
        for threshold in threshold:
            left_idx = X[:, feature] <= threshold
            right_idx = X[:, feature] > threshold
            if np.sum(left_idx) == 0 or np.sum(right_idx) == 0:
                continue
            gini_left = gini_impurity(y[left_idx])
            gini_right = gini_impurity(y[right_idx])
            weighted_gini = (len(y[left_idx]) * gini_left + len(y[right_idx]) * gini_right) / len(y)
            if weighted_gini < best_gini:
                best_gini = weighted_gini
                best_feature = feature
                best_threshold = threshold
    if best_feature is None:
            return Decisiontree(value=np.argmax(np.bincount(y)))
        
    left_idx = X[:, best_feature] <= best_threshold
    right_idx = X[:, best_feature] > best_threshold
    left = build_tree(X[left_idx], y[left_idx], depth + 1, max_depth)
    right = build_tree(X[right_idx], y[right_idx], depth + 1, max_depth)
    return Decisiontree(feature=best_feature, threshold=best_threshold, left=left, right=right)
```

```python
class Randomforest:
    def __init__(self, n_estimator=100, max_depth=5, max_features='sqrt'):
        self.n_estimator = n_estimator
        self.max_depth = max_depth
        self.max_feature = max_features
    def fit(self, X, y):
        self.trees = []
        nsamples = len(X_train)

        for _ in range(self.n_estimator):
            sample_indicies = np.random.choice(nsamples, nsamples, replace=True)

            X_sample = X_train[sample_indicies]
            Y_sample = y_train[sample_indicies]

            tree = build_tree(X_sample, Y_sample, max_depth=self.max_depth, max_feature=self.max_feature)
            self.trees.append(tree)
    
    def predict_one(self, tree, X):
        node = tree
        while node.value is None:
            if X[node.feature] <= node.threshold:
                node = node.left
            else:
                node = node.right
        return node.value

    def predict(self, X_test):
        predictions = np.array([self.predict_one(tree, x) for tree in self.tree for x in X_test])
        predictions = predictions.reshape(self.n_estimator, len(X_test))

        final_predictions = np.apply_along_axis(lambda x: np.argmax(np.bincount(x), 0, predictions))

        return final_predictions
```

```python
random_forest = Randomforest(n_estimator=100, max_depth=5, max_features='sqrt')
random_forest.fit(X_train, y_train)
```

```python
def accuracy(y_true, y_pred):
    correct = np.sum(y_true == y_pred)
    total = len(y_true)
    return correct / total
```

```python
range_depth = range(1,11)
best_accuracy = 0.0
best_depth = None
result = {}

for depth in range_depth:
    search = Randomforest(n_estimator=100, max_depth=depth, max_features='sqrt')
    search.fit(X_train, y_train)

    y_pred_test = search.predict(X_test)
    current_acc = accuracy(y_test, y_pred_test)
    result[depth] = current_acc
    if current_acc > best_accuracy:
        best_accuracy = best_accuracy
        best_depth = depth
    print(f"Depth: {depth} -- Akurasi: {current_acc}")
    
print(f"best depth: {best_depth} -- Akurasi: {best_accuracy}")
```

```python
range_tree = range(2,51)
best_tree = None
best_accuracy_tree = 0.0
result_tree = {}

for n_tree in range_tree:
    tree_search = Randomforest(n_estimator=n_tree, max_depth=best_depth, max_features='sqrt')

    tree_search.fit(X_train, y_train)
    y_pred_test_tree = tree_search.predict(X_test)
    current_acc_tree = accuracy(y_test, y_pred_test_tree)
    result_tree[n_tree] = current_acc_tree

    if current_acc_tree > best_accuracy_tree:
        best_accuracy_tree = current_acc_tree
        best_tree = n_tree
    print(f"Tree: {n_tree} -- Akurasi: {current_acc_tree}")
print(f"Best Tree: {best_tree} -- Akurasi: {best_accuracy_tree}")
```
