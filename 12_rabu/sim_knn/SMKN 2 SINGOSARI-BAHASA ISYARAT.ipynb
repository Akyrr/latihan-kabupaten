<b>B9. PCA (Principal Components Analysis)

```python
def fit_pca(X_train, n_components=2):
    mean = np.mean(X_train, axis=0)
    X_sentral = X_train - mean
    cov_matrix = np.cov(X_sentral, rowvar=False)
    eigenvalues, eigenvectors = np.linalg.eig(cov_matrix)
    sorted_indices = np.argsort(eigenvalues)[::-1]
    components = eigenvectors[:, sorted_indices[:n_components]]

    return mean, components

def transform_pca(X, mean, components):
    X_sentral = X - mean
    return np.dot(X_sentral, components)

mean, components = fit_pca(X_train, n_components=2)

X_train_pca = transform_pca(X_train, mean, components)
X_test_pca = transform_pca(X_test, mean, components)
```

# <b>C. Model 

<b>C1. Model KNN

```python
def hitung_jarak(x1, x2, metric='euclidean', p=3):
    if metric == 'euclidean':
        return np.linalg.norm(x1 - x2)
    elif metric == 'manhattan':
        return np.sum(np.abs(x1 - x2))
    elif metric == 'minkowski':
        return np.sum(np.abs(x1 - x2) ** p) ** (1 / p)
    
def knn(X_train, y_train, X_test, k=3, voting='majority', metric='euclidean', p=3, return_proba=False):
    pred = []
    probs = []

    for x_test in X_test:
        jarak = [hitung_jarak(x_test, x_train, metric, p) for x_train in X_train]
        k_indices = np.argsort(jarak)[:k]
        k_label = y_train[k_indices].astype(int)

        if voting == 'majority':
            label = np.bincount(k_label).argmax()
            proba = np.mean(k_label)
        else:
            weights = [1 / (jarak[i] + 1e-5) for i in k_indices]
            label = round(np.average(k_label, weights=weights))
            proba = np.average(k_label, weights=weights)

        pred.append(label)
        probs.append(proba)

    if return_proba:
        return np.array(pred), np.array(probs)
    
    return np.array(pred)

def akurasi(y_true, y_pred):
    return np.mean(y_true == y_pred)

```

<b>D1. Evaluasi Data Test

```python
def evaluasi_model(y_true, y_pred, title=""):

    num_classes = len(np.unique(y_true))
    conf_matrix = np.zeros((num_classes, num_classes), dtype=int)

    for true, pred in zip(y_true.astype(int), y_pred.astype(int)):
        conf_matrix[true, pred] += 1

    class_names = [f"Kelas {i + 1}" for i in range(num_classes)]

    tp = np.diag(conf_matrix)
    fp = np.sum(conf_matrix, axis=0) - tp
    fn = np.sum(conf_matrix, axis=1) - tp

    precision = tp / np.sum(conf_matrix, axis=0)
    recall = tp / np.sum(conf_matrix, axis=1)
    f1 = 2 * ((precision * recall) / (precision + recall))

    precision_macro = np.mean(precision)
    recall_macro = np.mean(recall)
    f1_macro = np.mean(f1)
    accuracy = np.sum(tp) / np.sum(conf_matrix)

    print(f"Evaluasi model {title}:")
    print(f"Precision : {precision_macro}")
    print(f"Recall : {recall_macro}")
    print(f"F1 score : {f1_macro}")
    print(f"Akurasi : {accuracy}")


evaluasi_model(y_test, y_pred, title="test")
```
